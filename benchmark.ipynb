Cílem benchmarku je vytvořit jednoduchou referenční úroveň pro predikci inflace v ČR.

Použity jsou dva přístupy:
1.Naivní baseline model
2.ednoduchý lineární regresní model

Benchmarky jsou vyhodnoceny pouze na validační sadě. Testovací sada zůstává stranou pro pozdější finální vyhodnocení.

IMPORT KNIHOVEN


```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error
```

NAČTENÍ DATASETŮ

```python
train = pd.read_csv("data/train.csv", parse_dates=["date"], index_col="date")
valid = pd.read_csv("data/validation.csv", parse_dates=["date"], index_col="date")
test = pd.read_csv("data/test.csv", parse_dates=["date"], index_col="date")

print("Train:", train.shape)
print("Validation:", valid.shape)
print("Test:", test.shape)
```

NAIVNÍ BASELNI

```python
y_valid = valid["target"]
y_pred_baseline = valid["baseline"]

rmse_baseline = np.sqrt(mean_squared_error(y_valid, y_pred_baseline))
mae_baseline = mean_absolute_error(y_valid, y_pred_baseline)

print("Baseline RMSE:", rmse_baseline)
print("Baseline MAE:", mae_baseline)
```

JEDNODUCHY ML BENCHMARK LINEÁRNÍ REGRESE

```python
features = ["infl_de", "rate", "oil", "gdp_growth"]

X_train = train[features]
y_train = train["target"]

X_valid = valid[features]
y_valid = valid["target"]

model = LinearRegression()
model.fit(X_train, y_train)

y_pred_lr = model.predict(X_valid)

rmse_lr = np.sqrt(mean_squared_error(y_valid, y_pred_lr))
mae_lr = mean_absolute_error(y_valid, y_pred_lr)

print("Linear Regression RMSE:", rmse_lr)
print("Linear Regression MAE:", mae_lr)
```

POROVNÁNÍ VÝSLEDNŮ

```python
results = pd.DataFrame({
    "model": ["Baseline", "Linear Regression"],
    "RMSE": [rmse_baseline, rmse_lr],
    "MAE": [mae_baseline, mae_lr]
})

results
```

## Vyhodnocení benchmarků

Naivní baseline model dosahuje lepších výsledků než lineární regresní model.

Baseline dosáhl RMSE = 0.762672, lineární regrese RMSE = 2.239225.

To naznačuje, že inflace má silnou setrvačnost a minulá hodnota inflace je velmi silným prediktorem budoucí hodnoty.

Lineární model využívající makroekonomické proměnné nedokázal tuto jednoduchou strategii překonat, což může být způsobeno omezeným množstvím dat a slabší predikční silou použitých proměnných.

Výsledek ukazuje, že baseline model představuje silnou referenční úroveň pro další modelování.
